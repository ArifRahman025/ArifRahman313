```python
# 1A: Bilangan ganjil 1-9
for angka in range(1, 10, 2):
    print(angka)

# 1B: Bilangan genap 2-50
for angka in range(2, 51, 2):
    print(angka, end=" ")
print()

# 1C: Hitung mundur 255 ke 0
for i in range(255, -1, -1):
    print(i, end=" ")
print()

# Tantangan: kelipatan 7 antara 1 dan 100
jumlah = 0
for angka in range(7, 101, 7):
    print(angka, end=" ")
    jumlah += 1
print(f"\nTotal kelipatan 7: {jumlah}")
```

```python
# 2A: while dasar
hitung = 1
while hitung <= 5:
    print(f"Iterasi ke-{hitung}")
    hitung += 1
print("Perulangan selesai.")

# 2B: Berhenti dengan input 99
kode_berhenti = 99
input_pengguna = 0

while input_pengguna != kode_berhenti:
    try:
        input_pengguna = int(input("Masukkan angka (99 untuk berhenti): "))
        if input_pengguna != kode_berhenti:
            print(f"  → Anda memasukkan: {input_pengguna}")
    except ValueError:
        print("  Input tidak valid. Harap masukkan angka.")
print("Perulangan dihentikan oleh nilai penghenti (99).")

# 2C: Berhenti dengan NIM (ganti dengan NIM Anda)
NIM = 4232501025
input_pengguna = 0

while input_pengguna != NIM:
    try:
        input_pengguna = int(input("Masukkan NIM untuk berhenti: "))
        if input_pengguna != NIM:
            print("  NIM belum cocok, coba lagi.")
    except ValueError:
        print("  Input tidak valid.")
print(f"NIM {NIM} cocok. Program berhenti.")

# 2D: Batas percobaan password
percobaan_maksimal = 3
percobaan_saat_ini = 1
password_benar = "polibatam"

while percobaan_saat_ini <= percobaan_maksimal:
    masukan = input(f"Percobaan ke-{percobaan_saat_ini}: Masukkan password: ")
    if masukan == password_benar:
        print("Akses diterima!")
        break
    else:
        print("  Password salah.")
    percobaan_saat_ini += 1
else:
    print("Kesempatan habis. Akses ditolak.")
```

```python
# 3A: Total belanja
harga_barang = [15000, 25000, 10000, 5000, 8000,
                12000, 30000, 7500, 22000, 18000, 9500, 4000]
total_belanja = 0
for harga in harga_barang:
    total_belanja += harga
print(f"Jumlah item  : {len(harga_barang)}")
print(f"Total belanja: Rp {total_belanja:,.0f}")

# 3B: Iterasi string
nama_institusi = "Politeknik Negeri Batam"
hasil_output = ""
for karakter in nama_institusi:
    hasil_output += karakter + "-"
print(hasil_output)

# 3C: enumerate()
bulan = ["Januari", "Februari", "Maret", "April", "Mei", "Juni",
         "Juli", "Agustus", "September", "Oktober", "November", "Desember"]
for i, nama in enumerate(bulan):
    print(f"Bulan ke-{i + 1:>2}: {nama}")

# 3D: Kelurahan di Batam
kelurahan_batam = [
    "Belian", "Baloi Permai", "Taman Baloi", "Sukajadi",
    "Batu Selicin", "Lubuk Baja Kota", "Kampung Pelita",
    "Batu Merah", "Tanjung Uma", "Sungai Jodoh", "Nongsa", "Sambau", "Muka Kuning"
]
print("Daftar Kelurahan di Kota Batam:")
print("-" * 35)
for i, nama in enumerate(kelurahan_batam):
    print(f"{i + 1:>3}. {nama}")
print(f"\nTotal: {len(kelurahan_batam)} kelurahan ditampilkan")
```

```python
# 4A: break saat menemukan -1
data_angka = [10, 20, 30, -1, 50, 60]
for angka in data_angka:
    if angka == -1:
        print("  → Angka -1 ditemukan! Proses dihentikan.")
        break
    print(f"  Memproses: {angka}")
print("Program selesai.")

# 4B: break saat sensor = 0
data_sensor = [45.2, 67.8, 89.1, 0, 55.3, 72.4]
for nilai in data_sensor:
    if nilai == 0:
        print("  ⚠ Pembacaan sensor = 0! Kemungkinan sensor mati.")
        break
    print(f"  Suhu terbaca: {nilai}°C")

# 4C: continue untuk data tidak valid
data_daya = [120.5, -1, 135.2, 0.0, 98.7, -2, 145.0]
for daya in data_daya:
    if daya <= 0:
        print(f"  (Melewati data tidak valid: {daya})")
        continue
    print(f"  Daya: {daya} MW")

# 4D: tambah berurutan sampai total > 20
total = 0
bilangan = 1
batas = 20
while total <= batas:
    total += bilangan
    bilangan += 1
print(f"Total pertama yang melebihi {batas}: {total}")
print(f"Bilangan terakhir ditambahkan: {bilangan - 1}")
```

```python
# 5A: "Python" 3 baris, 3 kata per baris
for baris in range(3):
    baris_teks = ""
    for _ in range(3):
        baris_teks += "Python "
    print(baris_teks.strip())

# 5B: Segitiga bintang
for baris in range(1, 6):
    for kolom in range(baris):
        print("*", end=" ")
    print()

# 5C: Tabel suhu boiler
nama_unit = ["Unit 1", "Unit 2", "Unit 3"]
suhu_harian = [
    [85.2, 87.1, 86.5, 88.0, 84.9, 85.7, 86.2],
    [90.1, 91.3, 89.7, 92.0, 90.5, 91.1, 90.8],
    [78.4, 79.2, 77.8, 80.1, 79.5, 78.9, 79.7],
]
hari = ["Sen", "Sel", "Rab", "Kam", "Jum", "Sab", "Min"]

print("Laporan Suhu Boiler (°C)")
print(f"{'Unit':<10}", end="")
for h in hari:
    print(f"{h:>6}", end="")
print(f"{'Rata2':>8}")
print("-" * 60)

for i, unit in enumerate(nama_unit):
    print(f"{unit:<10}", end="")
    total = 0
    for suhu in suhu_harian[i]:
        print(f"{suhu:>6.1f}", end="")
        total += suhu
    rata2 = total / len(suhu_harian[i])
    print(f"{rata2:>8.1f}")

# Tantangan: Tabel perkalian 1-9
print("\nTabel Perkalian")
print("    ", end="")
for j in range(1, 10):
    print(f"{j:>4}", end="")
print()
print("-" * 40)
for i in range(1, 10):
    print(f"{i:>3} |", end="")
    for j in range(1, 10):
        print(f"{i*j:>4}", end="")
    print()
```

```python
data_daya = [
    [142, 138, 145, 140],
    [115, 120, 118, 122],
    [ 95,  98,  92,  96],
]
nama_unit = ["Unit A", "Unit B", "Unit C"]

print("--- Soal L2: Rata-rata Daya per Unit ---")
for i in range(len(data_daya)):
    total = 0
    for daya in data_daya[i]:
        total += daya
    rata2 = total / len(data_daya[i])
    print(f"{nama_unit[i]}: {rata2:.1f} MW")
```

```python
import random
random.seed(42)

print("--- Soal L3: Monitoring Sensor ---")
siklus = 0
ada_anomali = True

while ada_anomali:
    siklus += 1
    ada_anomali = False
    print(f"\nSiklus ke-{siklus}:")
    for sensor_id in range(1, 5):
        suhu = random.randint(65, 95)
        if suhu < 70 or suhu > 90:
            print(f"  ⚠ Sensor {sensor_id}: {suhu}°C — ANOMALI")
            ada_anomali = True
        else:
            print(f"  ✓ Sensor {sensor_id}: {suhu}°C — Normal")

print(f"\nSemua sensor normal pada siklus ke-{siklus}.")
```
