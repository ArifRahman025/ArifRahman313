```python
# Data
jumlah_buku = 7
harga_buku = 18500
diskon_persen = 10
uang_budi = 150000
jumlah_orang = 4

# 1. Total belanja, diskon, dan jumlah bayar
total = jumlah_buku * harga_buku
diskon = total * diskon_persen // 100
bayar = total - diskon

# 2. Uang kembalian
kembalian = uang_budi - bayar

# 3. Patungan 4 orang (bilangan bulat)
per_orang = bayar // jumlah_orang
sisa = bayar % jumlah_orang

# 4. Apakah uang cukup?
cukup = uang_budi >= bayar

print("Total       :", total)
print("Diskon      :", diskon)
print("Bayar       :", bayar)
print("Kembalian   :", kembalian)
print("Per orang   :", per_orang)
print("Sisa        :", sisa)
print("Uang cukup? :", cukup)
```

```python
# Data
jumlah_panel = 12
daya_panel = 450          # Wp
jam_sinar = 5             # jam/hari
efisiensi = 0.8
kebutuhan = 25            # kWh/hari
tarif = 1444.70           # Rp/kWh
hari_per_bulan = 30

# 1. Energi harian (kWh)
energi_harian = jumlah_panel * daya_panel * jam_sinar * efisiensi / 1000

# 2. Energi bulanan dan nilai rupiah
energi_bulanan = energi_harian * hari_per_bulan
nilai_listrik = energi_bulanan * tarif

# 3. Kekurangan energi per hari
kekurangan = kebutuhan - energi_harian

# 4. Apakah mencukupi?
mencukupi = energi_harian >= kebutuhan

print("Energi harian (kWh) :", round(energi_harian, 2))
print("Energi bulanan (kWh):", round(energi_bulanan, 2))
print("Nilai listrik (Rp)  :", round(nilai_listrik, 2))
print("Kekurangan (kWh)    :", round(kekurangan, 2))
print("Mencukupi?          :", mencukupi)
```

```python
# Data
daya_generator = 200       # kW
jam_operasi = 8            # jam/hari
konsumsi = 0.25            # liter/kWh
harga_solar = 13500        # Rp/liter
kapasitas_tangki = 1000    # liter
arus = 20                  # A
hambatan = 0.5             # Ohm

# 1. Energi per hari (kWh)
energi = daya_generator * jam_operasi

# 2. Kebutuhan solar per hari dan biayanya
solar = energi * konsumsi
biaya = solar * harga_solar

# 3. Berapa hari penuh tangki cukup, dan sisa solar
hari_cukup = kapasitas_tangki // solar
sisa_solar = kapasitas_tangki % solar

# 4. Rugi daya kabel P = I^2 x R
rugi_kabel = arus ** 2 * hambatan

# 5. Apakah rugi kabel tinggi?
rugi_tinggi = rugi_kabel > 150 and arus >= 20

print("Energi (kWh)     :", energi)
print("Solar (liter)    :", solar)
print("Biaya (Rp)       :", biaya)
print("Tangki cukup     :", hari_cukup, "hari")
print("Sisa solar (L)   :", sisa_solar)
print("Rugi kabel (W)   :", rugi_kabel)
print("Rugi tinggi?     :", rugi_tinggi)
```
