```python
# Program 1: Menentukan Tahun Kabisat
# Ekspresi logika kompleks:
# (habis dibagi 4 DAN tidak habis dibagi 100) ATAU (habis dibagi 400)

def input_tahun():
    return int(input("Masukkan tahun: "))

def is_kabisat(tahun):
    return (tahun % 4 == 0 and tahun % 100 != 0) or (tahun % 400 == 0)

def tampilkan_hasil(tahun, kabisat):
    if kabisat:
        print(f"Tahun {tahun} adalah tahun kabisat")
    else:
        print(f"Tahun {tahun} BUKAN tahun kabisat")

tahun = input_tahun()
tampilkan_hasil(tahun, is_kabisat(tahun))
```

```python
# Program 2: Kalkulator Sederhana
# Lingkup (scope): variabel global & lokal
# Fungsi input, fungsi output, dan operasi aritmatika

riwayat = 0  # variabel GLOBAL (lingkup seluruh program)

def fungsi_input():
    a = float(input("Masukkan angka pertama : "))   # variabel LOKAL
    b = float(input("Masukkan angka kedua   : "))
    return a, b

def operasi_aritmatika(a, b):
    tambah = a + b
    kurang = a - b
    kali = a * b
    bagi = a / b if b != 0 else "tidak terdefinisi"
    sisa = a % b if b != 0 else "tidak terdefinisi"
    pangkat = a ** b
    return tambah, kurang, kali, bagi, sisa, pangkat

def fungsi_output(a, b, hasil):
    global riwayat
    riwayat += 1
    print("\n===== HASIL PERHITUNGAN =====")
    print(f"{a:g} + {b:g}  = {hasil[0]:g}")
    print(f"{a:g} - {b:g}  = {hasil[1]:g}")
    print(f"{a:g} x {b:g}  = {hasil[2]:g}")
    print(f"{a:g} / {b:g}  = {hasil[3] if isinstance(hasil[3], str) else format(hasil[3], 'g')}")
    print(f"{a:g} mod {b:g} = {hasil[4] if isinstance(hasil[4], str) else format(hasil[4], 'g')}")
    print(f"{a:g} ^ {b:g}  = {hasil[5]:g}")
    print(f"Perhitungan ke-{riwayat}")

x, y = fungsi_input()
fungsi_output(x, y, operasi_aritmatika(x, y))
```
