```python
energy = [
    5576, 6098, 6078, 6095, 6074, 6077, 5564, 5552, 6100, 6083,
    6097, 6089, 6066, 5533, 5566, 6097, 3592, 6059, 6100, 6098,
    5565, 5571, 6103, 6135, 6065, 6071, 6086, 5577, 5575, 6062,
]

# Jam operasi harian dari bagian "data 30 hari" di modulmu (30 nilai).
# Biarkan None jika belum ada; program tetap jalan tanpa langkah 2.
# Contoh: run_hours = [24, 24, 24, ..., 14, ..., 24]
run_hours = None

DAY = 17
i = DAY - 1      # hari 17 = indeks 16

# --- 1. Seberapa besar penurunannya? ---
prev = energy[i - 1]
change = (energy[i] - prev) / prev * 100
nxt = energy[i + 1]
rebound = (nxt - energy[i]) / energy[i] * 100
print(f"Hari {DAY}: {energy[i]} vs hari {DAY - 1}: {prev} -> {change:+.1f}%")
print(f"Hari {DAY + 1}: {nxt} ({rebound:+.1f}% dari hari {DAY}), pulih dalam sehari")

# --- 2. Cek jam operasi (alasan 1, 2 atau 3?) ---
print("\n== Cek run_hours ==")
if run_hours is None:
    print("run_hours belum diisi. Isi dari modulmu, lalu jalankan lagi.")
else:
    others = run_hours[:i] + run_hours[i + 1:]
    normal = sum(others) / len(others)
    ratio = run_hours[i] / normal
    print(f"run_hours[{i}] = {run_hours[i]}, rata-rata hari lain = {normal:.1f} ({ratio:.0%})")
    if ratio < 0.8:
        print("Jam operasi ikut turun -> condong ke gangguan/trip (1) atau perawatan/curtailment (2).")
        print("Bedakan dengan log alarm (trip) vs work order/jadwal dispatch (perawatan).")
    else:
        print("Jam operasi normal tetapi energi rendah -> curigai masalah meter/data (3).")
        print("Cek celah data per jam dan bandingkan dengan meter cadangan.")

# --- 3. Apakah ini sekadar efek akhir pekan? ---
# ASUMSI: hari 1 = Minggu (cocok dengan penurunan ~8% di hari 1, 7, 8, 14, 15, 21, 22, 28, 29).
# Cocokkan dengan blok weekday di modulmu.
def is_weekend(day):
    return (day - 1) % 7 in (0, 6)      # 0 = Minggu, 6 = Sabtu

weekday_vals = [energy[d - 1] for d in range(1, len(energy) + 1) if not is_weekend(d) and d != DAY]
weekend_vals = [energy[d - 1] for d in range(1, len(energy) + 1) if is_weekend(d)]
avg_weekday = sum(weekday_vals) / len(weekday_vals)
avg_weekend = sum(weekend_vals) / len(weekend_vals)

print("\n== Pola hari kerja vs akhir pekan ==")
print(f"Rata-rata hari kerja (tanpa hari {DAY}): {avg_weekday:,.0f}")
print(f"Rata-rata akhir pekan                : {avg_weekend:,.0f}")
print(f"Hari {DAY} jatuh pada hari kerja: {not is_weekend(DAY)}")
print(f"Hari {DAY} = {energy[i]:,} -> {energy[i] / avg_weekday - 1:+.1%} dari hari kerja biasa, "
      f"{energy[i] / avg_weekend - 1:+.1%} dari akhir pekan")
print("Kesimpulan: efek akhir pekan tidak menjelaskan hari 17; penyebabnya kejadian khusus.")

# --- 4. Tiga alasan dan data pembuktiannya ---
print("\n== Tiga alasan operasional ==")
print("1. Unit trip/gangguan   -> periksa run_hours[16], log alarm/trip, catatan operator")
print("2. Perawatan/curtailment -> periksa work order, jadwal dispatch, permintaan jaringan")
print("3. Masalah meter/data    -> periksa celah data per jam, meter cadangan, run_hours normal?")
```

```python
energy = [
    5576, 6098, 6078, 6095, 6074, 6077, 5564, 5552, 6100, 6083,
    6097, 6089, 6066, 5533, 5566, 6097, 3592, 6059, 6100, 6098,
    5565, 5571, 6103, 6135, 6065, 6071, 6086, 5577, 5575, 6062,
]
days = list(range(1, 31))
run_hours = [24] * 30
run_hours[16] = 18
DAY_NAMES = ["Sun", "Mon", "Tue", "Wed", "Thu", "Fri", "Sat"]

for i in range(len(energy)):
    if run_hours[i] < 24:
        print("Day", days[i], ":", run_hours[i], "h,", energy[i], "MWh")

weekday = []
weekend = []
for i in range(len(energy)):
    if DAY_NAMES[i % 7] == "Sat" or DAY_NAMES[i % 7] == "Sun":
        weekend.append(energy[i])
    else:
        weekday.append(energy[i])

print("Weekday average:", round(sum(weekday) / len(weekday)), "MWh")
print("Weekend average:", round(sum(weekend) / len(weekend)), "MWh")
```

```python
temps = [27, 29, 31, 30, 33, 32, 28]     # 7 suhu harian (nilai bebas)
print("temps awal:", temps)

# (1) Tiga hari pertama dan dua hari terakhir dengan slicing
print("\n(1) 3 hari pertama:", temps[:3])
print("    2 hari terakhir:", temps[-2:])

# (2) Tambah satu hari dengan append, lalu keluarkan lagi dengan pop
temps.append(35)
print("\n(2) setelah append(35):", temps)
removed = temps.pop()
print("    pop() mengeluarkan:", removed, "-> temps:", temps)

# (3) Versi terurut dengan sorted, temps tidak berubah
terurut = sorted(temps)
print("\n(3) terurut :", terurut)
print("    temps   :", temps, "(tidak berubah)")

# (4) Rata-rata, suhu tertinggi, dan harinya (hari pertama = 1)
average = sum(temps) / len(temps)
highest = max(temps)
day_of_highest = temps.index(highest) + 1
print(f"\n(4) rata-rata: {average:.2f}")
print(f"    tertinggi: {highest} pada hari ke-{day_of_highest}")

# (5) Daftar baru: suhu di atas rata-rata
above = []
for t in temps:
    if t > average:
        above.append(t)
print("\n(5) di atas rata-rata:", above)
```

```python
scores = [80, 65, 90, 75, 70, 85]

# (1) Skor pertama dan terakhir (indeks negatif untuk yang terakhir)
print("Skor pertama :", scores[0])
print("Skor terakhir:", scores[-1])

# (2) Rata-rata
average = sum(scores) / len(scores)
print("Rata-rata    :", average)

# (3) Berapa skor di atas rata-rata
count_above = 0
for s in scores:
    if s > average:
        count_above = count_above + 1
print("Di atas rata-rata:", count_above, "dari", len(scores))

# (4) Posisi skor terendah sebagai nomor siswa (siswa pertama = 1)
lowest_index = scores.index(min(scores))
print("Skor terendah:", min(scores), "-> siswa nomor", lowest_index + 1)

# (5) Daftar baru: setiap skor naik 5
boosted = []
for s in scores:
    boosted.append(s + 5)
print("Skor +5      :", boosted)
```

```python
energy = [
    5576, 6098, 6078, 6095, 6074, 6077, 5564, 5552, 6100, 6083,
    6097, 6089, 6066, 5533, 5566, 6097, 3592, 6059, 6100, 6098,
    5565, 5571, 6103, 6135, 6065, 6071, 6086, 5577, 5575, 6062,
]

n = len(energy)

# 1. Energi total dan rata-rata
total = sum(energy)
rata2 = total / n
print("=== LAPORAN PRODUKSI BULANAN (November 2026) ===")
print(f"Energi total     : {total:,} MWh")
print(f"Energi rata-rata : {rata2:,.1f} MWh/hari")

# 2. Tiga hari tertinggi dan tiga hari terendah (nomor hari + nilai)
pasangan = [(i + 1, energy[i]) for i in range(n)]

# Tertinggi: nilai menurun; jika sama, hari yang lebih awal didahulukan
tertinggi = sorted(pasangan, key=lambda p: (-p[1], p[0]))[:3]
# Terendah: nilai menaik; jika sama, hari yang lebih awal didahulukan
terendah = sorted(pasangan, key=lambda p: (p[1], p[0]))[:3]

print("\nTiga hari tertinggi:")
for hari, nilai in tertinggi:
    print(f"  Hari {hari:2d}: {nilai:,} MWh")

print("Tiga hari terendah:")
for hari, nilai in terendah:
    print(f"  Hari {hari:2d}: {nilai:,} MWh")

# 3. Jumlah hari di bawah target yang dimasukkan pengguna
while True:
    try:
        target = float(input("\nMasukkan target energi harian (MWh): "))
        break
    except ValueError:
        print("Input tidak valid, masukkan angka.")

hari_di_bawah = [i + 1 for i in range(n) if energy[i] < target]
print(f"Jumlah hari di bawah target ({target:g}): {len(hari_di_bawah)} hari")
print("Hari-harinya:", hari_di_bawah)

# 4. Hari ketika rata-rata bergerak 7 hari < 97% rata-rata bulanan
batas = 0.97 * rata2
print(f"\nRata-rata bergerak 7 hari di bawah 97% rata-rata bulanan (< {batas:,.1f}):")
ada = False
for i in range(6, n):                      # mulai hari ke-7 (indeks 6)
    jendela = energy[i - 6 : i + 1]        # 7 hari terakhir termasuk hari i
    avg7 = sum(jendela) / 7
    if avg7 < batas:
        print(f"  Hari {i + 1:2d}: {avg7:,.1f} MWh/hari")
        ada = True
if not ada:
    print("  Tidak ada.")

# 5. Total energi per minggu (minggu 1-4 + sisa hari)
print("\nTotal energi per minggu:")
for m in range(4):
    awal = m * 7
    akhir = awal + 7
    print(f"  Minggu {m + 1} (hari {awal + 1}-{akhir}): {sum(energy[awal:akhir]):,} MWh")

sisa = energy[28:]
if sisa:
    print(f"  Sisa hari (hari 29-{n}): {sum(sisa):,} MWh")
```
