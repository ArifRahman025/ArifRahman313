```python
suhu = 15
cuaca = "salju"
if suhu >= 10 and cuaca == "dingin":
  print("Harus memakai jaket tebal untuk pergi keluar.")
elif suhu <= 10 and cuaca == "dingin":
  print("Harus mmakai jaket ringan untuk pergi keluar.")
elif cuaca == "salju":
  print("sebaiknya tidak pergi keluar.")
else:
  print("informasi cuaca tidak pasti.")
```

```python
# Program untuk mengecek gratis ongkir dan diskon
total_belanja = int(input("Total belanja: Rp "))

if total_belanja > 150000:
    diskon = total_belanja * 0.15
    total_bayar = total_belanja - diskon
    print("Anda mendapat diskon 15% dan gratis ongkir! Total bayar: Rp {total_bayar}")
else:
    ongkir = 10000
    total_bayar = total_belanja + ongkir
    print("Total bayar (termasuk ongkir Rp {ongkir}): Rp {total_bayar}")
```

```python
celsius = float(input("Masukkan suhu (°C): "))
print("1. Fahrenheit (°F)")
print("2. Kelvin (K)")
print("3. Reamur (°R)")
pilih = int(input("Pilih konversi (1/2/3): "))

if pilih == 1:
    print("Dalam Fahrenheit:", (celsius * 9 / 5) + 32)
elif pilih == 2:
    print("Dalam Kelvin:", celsius + 273.15)
elif pilih == 3:
    print("Dalam Reamur:", celsius * 4 / 5)
else:
    print("Pilihan tidak valid")
```
