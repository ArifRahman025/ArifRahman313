```python
temps = [28, 31, 30, 33, 29]

hot = []                          # the new list starts empty
for t in temps:
    if t >= 30:
        hot.append(t)
print("Hot values  :", hot)
print("How many    :", len(hot))

# To keep the day numbers instead of the values, loop by position
hot_days = []
for i in range(len(temps)):
    if temps[i] >= 30:
        hot_days.append(i + 1)
print("Hot days    :", hot_days)

# Optional: the same filter as a list comprehension (a compact loop)
hot_again = [t for t in temps if t >= 30]
print("Same result?", hot_again == hot)
```
