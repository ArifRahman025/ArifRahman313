```python
import builtins
pembangkit = [
    {"nama": "Unit-1", "kapasitas_Mw": 100, "status": "operasi"},
    {"nama": "Unit-2", "kapasitas_Mw": 1500, "status": "operasi"},
    {"nama": "Unit-3", "kapasitas_Mw": 80, "status": "operasi"}
]

# Cetak nama unit kedua
builtins.print(pembangkit[1]["nama"])

# Ubah status salah satu unit
pembangkit[2]["status"] = "operasi"

# Tambahkan satu unit baru
pembangkit.append({
    "nama": "Unit-4",
    "kapasitas_Mw": 120,
    "status": "operasi"
})

# Cetak seluruh data
builtins.print(pembangkit)
```
