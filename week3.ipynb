```python
#RPE311 DASAR PEMROGRAMAN
#Pertemuan 3 - Expressions & Operators
#Nama : Arif Rahman
#NIM : 4232501025
#Tanggal : 21 September 2026
#Tujuan : Menghitung hasil operasi aritmatika (+,-,*,/,//,%) dari dua bilangan bulat yang dimasukkan oleh pengguna.

#Soal 1
#Operasi aritmatika

print("SOAL 1-OPERASI ARITMATIKA")

bilangan1 = int(input("Masukkan bilangan pertama(a):"))
bilangan2 = int(input("Masukkan bilangan kedua (b):"))

penjumlahan = bilangan1 + bilangan2
pengurangan = bilangan1 - bilangan2
perkalian = bilangan1 * bilangan2
pembagian = bilangan1 / bilangan2
pembagian_bulat = bilangan1 // bilangan2
modulus = bilangan1 % bilangan2

print(f"a + b = {penjumlahan}")
print(f"a - b = {pengurangan}")
print(f"a * b = {perkalian}")
print(f"a / b = {pembagian}")
print(f"a // b = {pembagian_bulat}")
print(f"a % b = {modulus}")
```

```python
#Nama : Arif Rahman
#NIM : 4232501025
#Tanggal : 21 September 2026
#Tujuan : Menentukan apakah seseorang boleh mengemudi sendiri menggunakan operator logika 'and', sekaligus latihan operator precedence dengan tanda kurung.

umur = int(input("Masukkan umur:"))
punya_sim = input("Punya SIM(True/False):").strip().lower()== "true"

# Operator logika 'and' sederhana
boleh_mengemudi_sendiri = (umur >=17) and punya_sim
print(f"Boleh mengemudi sendiri: {boleh_mengemudi_sendiri}")

# Latihan operator precedence
# 'and' memiliki prioritas lebih tinggi daripada 'or', sehingga
# tanpa tanda kurung, ekspresi berikut bisa salah dibaca.
# Maksud kita: (umur remaja 15-16 TAHUN) ATAU (sudah punya SIM)
# dianggap boleh mengemudi dengan pengawasan.
boleh_dengan_pengawasan = (15 <= umur < 17)or punya_sim
print(f"Boleh mengemudi dengan pengawasan: {boleh_dengan_pengawasan}")

# Contoh jika tanda kurung dilupakan, hasilnya bisa berbeda maknanya:
# umur >= 15 and umur < 17 or punya_sim
# (tanpa kurung, ini tetap dievaluasi and dulu baru or, tapi kurung
# eksplisit membuat maksud ekspresi lebih jelas dan aman dari kesalahan logika)
```

```python
#Nama : Arif Rahman
#NIM : 4232501025
#Tanggal : 21 September 2026
#Tujuan : Menghitung efisiensi termal unit pembangkit efisiensi (%)=(energi output/input)*100

energi_input = float(input("Masukkan energi input(MJ):"))
energi_output = float(input("Masukkan energi output(MJ):"))

# Tanda kurung wajib di sini agar pembagian dilakukan lebih dulu
# sebelum dikalikan 100 (sesuai operator precedence)
efisiensi = (energi_output/energi_input)*100

print(f"Efisiensi termal unit pembangkit: {efisiensi:.2f}%")
```
