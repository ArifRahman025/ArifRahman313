```python
# Soal 1. Daya Generator dan Efisiensi Turbin

# a. Simpan tegangan dan arus, hitung daya listrik
tegangan = 220   # Volt
arus = 15        # Ampere
daya = tegangan * arus   # P = V × I

print(f"Daya listrik = {daya} W")

# b. Hitung efisiensi turbin
energi_input = 500
energi_output = 400
efisiensi = (energi_output / energi_input) * 100

print(f"Efisiensi = {efisiensi:.2f}%")
```

- Bagian a: nilai tegangan (220 V) dan arus (15 A) disimpan dalam variabel, lalu daya dihitung dengan rumus P = V × I = 3300 W dan ditampilkan memakai f-string.
- Bagian b: efisiensi dihitung dengan (energi_output / energi_input) * 100. Format :.2f pada f-string membuat hasilnya tampil dengan 2 angka desimal (80.00).

```python
# Soal 3. Data Suhu Boiler

# Data suhu pada 5 titik boiler (°C), disimpan dalam list
suhu = [92.5, 88.7, 95.1, 102.3, 97.8]

# Bagian a: Suhu pada titik ke-3
# Indeks list dimulai dari 0, jadi titik ke-3 berada di indeks 2
print(f"a. Suhu pada titik ke-3 = {suhu[2]} °C")

# Bagian b: Tiga pembacaan pertama dengan slicing
# Slicing [0:3] mengambil indeks 0, 1, 2 (indeks 3 tidak ikut)
print(f"b. Tiga pembacaan pertama = {suhu[0:3]}")

# Bagian c: Tambah pembacaan baru, lalu hitung jumlah data
# append() menambahkan elemen baru di akhir list
suhu.append(99.4)

# len() menghitung jumlah elemen di dalam list
print(f"c. Jumlah data setelah penambahan = {len(suhu)}")
print(f"   Isi list sekarang = {suhu}")
```

- Bagian a: Python memakai indeks yang dimulai dari 0, sehingga titik ke-3 diakses dengan suhu[2], bukan suhu[3]. Hasilnya 95.1.
- Bagian b: Slicing suhu[0:3] mengambil elemen dari indeks 0 sampai sebelum indeks 3, yaitu tiga data pertama. Boleh juga ditulis suhu[:3].
- Bagian c: Nilai baru 99.4 (angka yang tertera di soal) ditambahkan dengan append(), lalu len(suhu) menghitung jumlah datanya, sehingga menjadi 6.

```python
# Soal 5. Katup Pengaman

# tekanan_uap dibaca dari input (bertipe float)
# float() mengubah teks input menjadi bilangan desimal
tekanan_uap = float(input("Masukkan tekanan uap (psi): "))

# status_manual_override diberikan langsung dalam kode (bertipe bool)
# Ubah nilainya menjadi True atau False untuk menguji kombinasi lain
status_manual_override = False

# Menentukan status katup dengan if-elif-else dan operator logika
# Urutan pengecekan penting: syarat manual override diperiksa lebih dulu

if tekanan_uap > 150 and not status_manual_override:
    # Tekanan lebih dari 150 psi DAN override bernilai False
    print("Katup Terbuka")
elif status_manual_override:
    # Override bernilai True (berapa pun tekanannya)
    print("Katup Mode Manual")
else:
    # Selain dua kondisi di atas
    print("Katup Tertutup")

```

- tekanan_uap dibaca lewat input() lalu diubah ke float, sedangkan status_manual_override ditulis langsung di kode sebagai True atau False.
- Kondisi pertama tekanan_uap > 150 and not status_manual_override bernilai benar hanya jika tekanan lebih dari 150 (bukan sama dengan) dan override False. Karena itu tekanan 150 menghasilkan “Katup Tertutup”.
- Kondisi kedua elif status_manual_override menangkap semua kasus override True, sehingga hasilnya selalu “Katup Mode Manual” berapa pun tekanannya.
- else menangani sisanya, yaitu tekanan 150 psi atau kurang dengan override False, sehingga hasilnya “Katup Tertutup”.

```python
# Soal 7. Pemanas Mencapai Target

# Suhu awal dan target suhu pemanas (°C)
suhu_saat_ini = 60
target = 85

# Perulangan while berjalan selama suhu belum mencapai target
while suhu_saat_ini < target:
    # Naikkan suhu 5 derajat setiap putaran
    suhu_saat_ini += 5

    # Cetak suhu setelah dinaikkan
    print(f"Suhu saat ini: {suhu_saat_ini} C")

    # Saat suhu tepat 80, cetak peringatan
    if suhu_saat_ini == 80:
        print("WARNING: Mendekati target!")

# Setelah loop selesai, cetak pesan target tercapai
print(f"Target {target} C tercapai.")
```

- Nilai awal: suhu_saat_ini = 60 dan target = 85.
Perulangan while suhu_saat_ini < target: loop terus berjalan selama suhu masih kurang dari 85. Begitu suhu mencapai 85, kondisi menjadi salah dan loop berhenti.
- suhu_saat_ini += 5: suhu dinaikkan 5 derajat, lalu dicetak setelah dinaikkan. Itulah sebabnya output dimulai dari 65, bukan 60.
- if suhu_saat_ini == 80: peringatan dicetak tepat setelah baris “Suhu saat ini: 80 C”, sesuai contoh output. Karena if berada di dalam loop, pengecekannya dilakukan di setiap putaran.
- Setelah loop: pesan “Target 85 C tercapai.” dicetak sekali saja karena berada di luar loop (tanpa indentasi).

```python
# Soal 9. Pembacaan Vibrasi dengan Sensor Rusak

# Data vibrasi lima unit mesin dibaca sebagai teks (satu sensor rusak: "ERR")
data = ["1.2", "0.9", "ERR", "3.1", "1.8"]

# Penghitung jumlah data yang tidak valid
jumlah_tidak_valid = 0

# enumerate(data, start=1) memberi nomor unit yang dimulai dari 1
for nomor, teks in enumerate(data, start=1):

    # Bagian a: Ubah teks menjadi float
    try:
        nilai = float(teks)   # "ERR" akan memicu ValueError
    except ValueError:
        print(f"Unit {nomor}: data tidak valid")
        jumlah_tidak_valid += 1   # tambahkan ke penghitung data tidak valid
        continue                  # lanjut ke unit berikutnya

    # Bagian b: Nilai melebihi 2.0
    if nilai > 2.0:
        print(f"Warning: Unit {nomor} vibrasi tinggi!")

        # Bagian c: Nilai melebihi 3.0
        # Warning sudah dicetak di atas, lalu cetak KRITIS dan hentikan loop
        if nilai > 3.0:
            print("KRITIS: pemeriksaan dihentikan")
            break
```

- enumerate(data, start=1): menghasilkan nomor unit (mulai dari 1) sekaligus teks datanya, jadi tidak perlu variabel penghitung terpisah.
- Bagian a (try-except ValueError): float("ERR") gagal dan memicu ValueError. Saat itu terjadi, program mencetak “data tidak valid”, menambah jumlah_tidak_valid, lalu continue melompati sisa kode dan lanjut ke unit berikutnya.
- Bagian b: jika nilai > 2.0, cetak Warning: Unit [n] vibrasi tinggi!. Unit 1 (1.2), unit 2 (0.9), dan unit 5 (1.8) tidak melebihi 2.0, sehingga tidak ada peringatan.
- Bagian c: if nilai > 3.0 berada di dalam blok if nilai > 2.0, sehingga Warning selalu tercetak lebih dulu, baru kemudian “KRITIS” dan break menghentikan perulangan. Pada data ini, unit 4 bernilai 3.1 sehingga loop berhenti di situ dan unit 5 tidak diperiksa.
- Bagian d: print terakhir berada di luar loop (tanpa indentasi), sehingga tetap dijalankan setelah loop selesai, baik karena break maupun karena data habis.

```python
# Challenge: Pemantauan Beban Listrik Harian

# MASALAH:
# Sebuah gardu memantau beban listrik (kW) setiap jam. Data dibaca
# sebagai teks dari sensor, dan sebagian bisa rusak ("ERR").
# Program harus: memvalidasi data, mengklasifikasikan tingkat beban,
# menghitung statistik, dan menghitung perkiraan biaya listrik.
#
# CARA KERJA:
# 1. Data disimpan dalam list, lalu diperiksa satu per satu dengan for.
# 2. try-except menangani data rusak (tidak bisa diubah ke float).
# 3. if-elif-else mengklasifikasikan beban: Normal, Tinggi, Kritis.
# 4. Data valid dikumpulkan ke list baru untuk dihitung statistiknya.
# 5. Biaya dihitung dari total energi (kWh) x tarif per kWh.


# Data dan konstanta
# Beban per jam (kW) dari sensor, bertipe teks. Data pertama jam 00:00.
data_beban = ["320", "295", "310", "ERR", "450", "680", "720", "ERR",
              "540", "610", "830", "470", "390", "350", "300", "280"]

BATAS_TINGGI = 600     # kW, di atas ini dianggap beban tinggi
BATAS_KRITIS = 800     # kW, di atas ini dianggap beban kritis
TARIF_PER_KWH = 1444.7 # Rupiah per kWh

# Variabel penampung hasil
beban_valid = []       # list untuk menyimpan data yang valid
jumlah_error = 0       # penghitung data sensor rusak
jam_tinggi = []        # list jam dengan beban tinggi
jam_kritis = []        # list jam dengan beban kritis

# Proses pemeriksaan setiap data
# enumerate memberi nomor jam (0-15) sekaligus teks datanya
for jam, teks in enumerate(data_beban):

    # try-except: tangani data sensor rusak
    try:
        beban = float(teks)        # "ERR" akan memicu ValueError
    except ValueError:
        print(f"Jam {jam:02d}:00 -> data sensor rusak, dilewati")
        jumlah_error += 1
        continue                   # lanjut ke jam berikutnya

    # Data valid disimpan ke list
    beban_valid.append(beban)

    # if-elif-else: klasifikasi tingkat beban
    if beban > BATAS_KRITIS:
        status = "KRITIS"
        jam_kritis.append(jam)
    elif beban > BATAS_TINGGI:
        status = "TINGGI"
        jam_tinggi.append(jam)
    else:
        status = "Normal"

    print(f"Jam {jam:02d}:00 -> {beban:7.1f} kW ({status})")

# Perhitungan statistik
# Cek dulu agar tidak terjadi pembagian dengan nol bila semua data rusak
if len(beban_valid) > 0:
    beban_maks = max(beban_valid)
    beban_min = min(beban_valid)
    beban_rata = sum(beban_valid) / len(beban_valid)

    # Jam beban puncak: cari indeks nilai maksimum pada data asli
    jam_puncak = data_beban.index(str(int(beban_maks)))

    # Asumsi tiap data mewakili 1 jam, maka energi (kWh) = jumlah beban (kW)
    total_energi = sum(beban_valid)
    biaya = total_energi * TARIF_PER_KWH

    # Tampilkan laporan
    print("\n" + "=" * 45)
    print("LAPORAN BEBAN LISTRIK HARIAN")
    print("=" * 45)
    print(f"Jumlah data valid  : {len(beban_valid)}")
    print(f"Jumlah data rusak  : {jumlah_error}")
    print(f"Beban maksimum     : {beban_maks:.1f} kW (jam {jam_puncak:02d}:00)")
    print(f"Beban minimum      : {beban_min:.1f} kW")
    print(f"Beban rata-rata    : {beban_rata:.2f} kW")
    print(f"Jam beban tinggi   : {jam_tinggi}")
    print(f"Jam beban kritis   : {jam_kritis}")
    print(f"Total energi       : {total_energi:.1f} kWh")
    print(f"Perkiraan biaya    : Rp {biaya:,.0f}")

    # Peringatan akhir berdasarkan ada tidaknya beban kritis
    if len(jam_kritis) > 0:
        print("\nPERINGATAN: Terdapat beban kritis, periksa kapasitas gardu!")
    else:
        print("\nStatus: Beban gardu masih dalam batas aman.")
else:
    print("Tidak ada data valid yang dapat dihitung.")
```

- Masalah: memantau beban listrik per jam, mendeteksi jam beban tinggi/kritis, dan memperkirakan biaya, sambil menangani data sensor yang rusak.
- list: data_beban menyimpan data mentah, sedangkan beban_valid, jam_tinggi, dan jam_kritis menyimpan hasil pemrosesan.
- for + enumerate: memeriksa setiap data sekaligus mendapatkan nomor jam.
- try-except ValueError: data “ERR” tidak membuat program berhenti, tetapi dicatat sebagai rusak lalu dilewati dengan continue.
- if-elif-else: mengklasifikasikan beban menjadi Normal, TINGGI (> 600 kW), atau KRITIS (> 800 kW). Urutan pengecekan dari batas tertinggi penting agar nilai kritis tidak salah masuk ke kategori tinggi.
- Konstanta (BATAS_TINGGI, dll.): ditulis dengan huruf kapital agar mudah diubah tanpa mengutak-atik logika program.
- Pengecekan len(beban_valid) > 0: mencegah error pembagian dengan nol jika semua data rusak.
- Format f-string: :7.1f untuk rata kanan dengan 1 desimal, :,.0f untuk pemisah ribuan pada biaya.
